```python
from pathlib import Path

import warnings

import numpy as np

import pandas as pd

import matplotlib.pyplot as plt

import statsmodels.api as sm

from scipy.special import expit

from scipy.stats import t as student_t

from sklearn.metrics import roc_auc_score, roc_curve, log_loss

from sklearn.model_selection import StratifiedGroupKFold

from statsmodels.stats.outliers_influence import variance_inflation_factor

from IPython.display import display

FEATURES = ["WC/TA", "RE/TA", "EBIT/TA", "ME/TL", "S/TA"]

MIN_GINI = 0.08

MAX_CORRELATION = 0.70

MAX_VIF = 5.0

MAX_P_VALUE = 0.05

pd.set_option("display.max_columns", None)
```

Здесь мы для каждого фактора посмотрели описательные статистики: среднее, std, min/max, основные квантили.
отдельно посмотрели выбросы по IQR как описано в положении. Выбросы мы не трогали в конечном счете, потому что они имели экономический смысл.
Еще я посчитал со сколькими выбросами в каждом факторе ассоциированы дефолты.

```python
descriptive = df[FEATURES].describe(percentiles=[.01, .25, .5, .75, .99]).T

iqr_rows = []

for col in FEATURES:

    q1, q3 = df[col].quantile([.25, .75])

    low, high = q1 - 1.5 * (q3 - q1), q3 + 1.5 * (q3 - q1)

    flag = (df[col] < low) | (df[col] > high)

    iqr_rows.append({"фактор": col, "нижняя граница": low, "верхняя граница": high,
                     "IQR-наблюдений": int(flag.sum()), "среди них дефолтов": int(df.loc[flag, "Default"].sum())})

display(descriptive.drop(columns="count").round(4))

display(pd.DataFrame(iqr_rows).set_index("фактор"))
```

Здесь я задал вспомогательные функции для обучения логистической регрессии.

design() формирует матрицу признаков и добавляет константу(свободный член модели)

fit_logit() обучает логистическую регрессию для выбранного набора факторов. Стандартные ошибки считаются как cluster robust по ID, чтобы учитывать зависимость наблюдений одной и той же компании в разные годы. Также отдельно проверяется, что алгоритм обучения действительно сошёлся.

```python
def design(frame, columns):

    return sm.add_constant(frame[list(columns)], has_constant="add")


def fit_logit(frame, columns):

    result = sm.Logit(frame["Default"], design(frame, columns)).fit(

        disp=False, maxiter=200, cov_type="cluster",

        cov_kwds={"groups": frame["ID"]},

    )

    if not result.mle_retvals["converged"]:

        raise RuntimeError(f"Logit не сошёлся: {columns}")

    return result
```

Здесь я задал процедуру отбора факторов, которая потом применяется отдельно на каждом train-фолде и в конце на всей выборке.

Сначала для каждого фактора считается направленный однофакторный Gini. Факторы с Gini ниже 8% или с направлением связи, противоречащим экономическому смыслу, исключаются.

Дальше среди оставшихся факторов проверяется взаимная корреляция по Пирсону и Спирмену. Если максимальная корреляция в паре превышает 0.7, удаляется более слабый фактор по однофакторному Gini.

После этого проверяется мультиколлинеарность по VIF. Если у факторов VIF выше 5, из проблемных факторов также исключается более слабый по Gini.

На последнем этапе на оставшихся факторах строится совместный Logit. Последовательно исключаются факторы с неправильным знаком коэффициента или с cluster-robust p-value > 0.05, после каждого исключения модель оценивается заново.

Функция возвращает итоговый набор факторов, обученную модель и историю причин, по которым остальные факторы были исключены.

```python
def fit_score_model(train):
    """Каждый отбор выполняется только по переданному train."""
    train = train.reset_index(drop=True)
    y = train["Default"]
    if y.nunique() != 2:
        raise ValueError("Для оценки Logit нужны оба класса в train.")
    screen = pd.DataFrame(index=FEATURES, columns=["signed_gini", "status"], dtype=object)
    history = []

    # Для этих пяти отношений более высокое значение заранее трактуем как меньший риск.
    for col in FEATURES:
        screen.loc[col, "signed_gini"] = float(1 - 2 * roc_auc_score(y, train[col]))
    selected = [col for col in FEATURES if screen.loc[col, "signed_gini"] > MIN_GINI]
    for col in FEATURES:
        if col not in selected:
            history.append((col, "Gini < 0.08 или обратный экономический знак"))
    if not selected:
        raise RuntimeError("После однофакторного отбора не осталось признаков.")

    # Два критерия корреляции заранее зафиксированы; исключается более слабый фактор.
    while len(selected) > 1:
        pears = train[selected].corr(method="pearson").abs()
        spears = train[selected].corr(method="spearman").abs()
        pairs = [(a, b, max(pears.loc[a, b], spears.loc[a, b]))
                 for i, a in enumerate(selected) for b in selected[i + 1:]]
        if not pairs or max(v for _, _, v in pairs) <= MAX_CORRELATION:
            break
        a, b, strength = max(pairs, key=lambda item: item[2])
        weak = min((a, b), key=lambda name: (screen.loc[name, "signed_gini"], name))
        selected.remove(weak)
        history.append((weak, f"корреляция {strength:.3f} с {b if weak == a else a}"))

    while len(selected) > 1:
        x = design(train, selected).to_numpy()
        vifs = {col: float(variance_inflation_factor(x, i + 1))
                for i, col in enumerate(selected)}
        if max(vifs.values()) <= MAX_VIF:
            break
        candidate = [c for c, v in vifs.items() if v > MAX_VIF]
        weak = min(candidate, key=lambda name: (screen.loc[name, "signed_gini"], name))
        history.append((weak, f"VIF {vifs[weak]:.2f} > {MAX_VIF:g}"))
        selected.remove(weak)

    # Последовательно удаляем худший неверный знак или p > 0.05.
    while True:
        model = fit_logit(train, selected)
        wrong = [c for c in selected if model.params[c] >= 0]
        if wrong:
            weakest = max(wrong, key=lambda c: model.params[c])
            reason = "неверный знак коэффициента"
        else:
            worst = max(selected, key=lambda c: model.pvalues[c])
            weakest = worst if model.pvalues[worst] > MAX_P_VALUE else None
            reason = f"cluster-robust p={model.pvalues[worst]:.4f} > {MAX_P_VALUE:.2f}"
        if weakest is None:
            break
        if len(selected) == 1:
            raise RuntimeError("Даже единственный оставшийся фактор нарушает правило знака или p-value.")
        selected.remove(weakest)
        history.append((weakest, reason))

    for col in FEATURES:
        screen.loc[col, "status"] = "в модели" if col in selected else "исключён"

    return {"features": selected, "model": model, "screen": screen, "history": history}
```

Здесь заданы две вспомогательные функции.

linear_score() рассчитывает линейный скор логистической модели (эта греческая) для заданных наблюдений. Сам PD получается позднее применением логистической функции к этому скору.

make_folds() формирует разбиения для cross-validation с помощью StratifiedGroupKFold. Разделение проводится по ID, чтобы одна компания не попадала одновременно в train и test, как я ранее упомянал. Также проверяется, что в обеих частях каждого разбиения присутствуют как дефолтные, так и недефолтные наблюдения.

В конце через yield функция по очереди отдаёт очередную пару индексов train и test для каждого фолда.

```python
def linear_score(fitted, frame):
    return np.asarray(design(frame, fitted["features"]) @ fitted["model"].params, dtype=float)


def make_folds(data, n_splits, random_state):
    cv = StratifiedGroupKFold(n_splits=n_splits, shuffle=True, random_state=random_state)
    for tr, te in cv.split(data[FEATURES], data["Default"], groups=data["ID"]):
        assert not (set(data.iloc[tr]["ID"]) & set(data.iloc[te]["ID"]))
        assert data.iloc[tr]["Default"].nunique() == 2
        assert data.iloc[te]["Default"].nunique() == 2
        yield tr, te
```

Здесь реализована кросс-валидация всей процедуры построения модели.

Для каждого из пяти фолдов:

1)формируются train и test;

2)только на train заново выполняется весь отбор факторов;

3)на выбранных факторах обучается логистическая регрессия;

4)для test рассчитывается линейный score, после чего через логистическую функцию получается PD;

5)полученный прогноз сохраняется как PD_OOF.

Дополнительно для каждого фолда сохраняются размер train и test, число дефолтов, выбранные факторы и AUC.

Таким образом, каждый PD_OOF получен моделью, которая не обучалась на соответствующем наблюдении. Это позволяет оценивать качество всей процедуры отбора и обучения без использования test при построении модели.

```python
def validate_pipeline(data, outer_folds=5, random_state=2026):
    """CV оценивает всю процедуру отбора факторов и обучения Logit."""
    data = data.reset_index(drop=True)
    output = pd.DataFrame(
        index=data.index,
        columns=["outer_fold", "PD_OOF"],
        dtype=float,
    )
    records = []

    for fold, (tr, test) in enumerate(make_folds(data, outer_folds, random_state), 1):
        training = data.iloc[tr].reset_index(drop=True)
        testing = data.iloc[test]

        fitted = fit_score_model(training)
        score_test = linear_score(fitted, testing)
        pd_test = expit(score_test)

        output.loc[test, "outer_fold"] = fold
        output.loc[test, "PD_OOF"] = pd_test

        records.append({
            "fold": fold,
            "train_n": len(tr),
            "test_n": len(test),
            "train_defaults": int(training["Default"].sum()),
            "test_defaults": int(testing["Default"].sum()),
            "selected": ", ".join(fitted["features"]),
            "AUC": roc_auc_score(testing["Default"], pd_test),
        })

    assert output.notna().all().all()
    return output, pd.DataFrame(records)
```

Здесь запускается кросс-валидация на 5 фолдах.

Для каждого фолда выводятся размер train и test, число дефолтов в них, выбранные факторы и AUC на test-части.

Дополнительно считается, сколько раз каждый фактор был выбран в пяти фолдах. Это позволяет посмотреть, насколько устойчив отбор факторов к изменению состава обучающей выборки.

AUC на отдельных фолдах находился примерно в диапазоне от 0.745 до 0.954, то есть качество модели оставалось выше случайного, однако заметно менялось в зависимости от состава выборки.

```python
oof, folds = validate_pipeline(df, outer_folds=5, random_state=2026)

display(folds[["fold", "train_n", "test_n", "train_defaults", "test_defaults", "selected", "AUC"]].round(4))

selection_frequency = pd.DataFrame({
    "фактор": FEATURES,
    "внешних фолдов из 5": [
        folds["selected"].str.split(", ").map(lambda s: col in s).sum()
        for col in FEATURES
    ],
})

display(selection_frequency)
```

Здесь оценивается качество модели на всех OOF-прогнозах и строятся графики. 
Сначала по PD_OOF рассчитываются основные показатели дискриминационной способности:

1)AUC;

2)Gini, рассчитанный как 2*AUC - 1;

3)KS как максимальная разница между True Positive Rate и False Positive Rate.

На первом графике строится ROC-кривая по всем OOF-прогнозам, а на втором — AUC отдельно по каждому фолду. Пунктирная линия на уровне 0.5 соответствует качеству случайного ранжирования.

```python
y = df["Default"].to_numpy()
pd_oof = oof["PD_OOF"].to_numpy(dtype=float)

fpr, tpr, _ = roc_curve(y, pd_oof)
auc = roc_auc_score(y, pd_oof)
ks = np.max(tpr - fpr)

print(f"Внешние OOF: AUC={auc:.4f}; Gini={2*auc-1:.2%}; KS={ks:.4f}")

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].plot(fpr, tpr, label=f"AUC={auc:.3f}")
axes[0].plot([0, 1], [0, 1], "--", color="gray")
axes[0].set(xlabel="False positive rate", ylabel="True positive rate", title="ROC: внешние OOF прогнозы")
axes[0].legend()

axes[1].bar(folds.fold.astype(str), folds.AUC, color="#365f9c")
axes[1].axhline(.5, ls="--", color="gray")
axes[1].set(xlabel="Внешний фолд", ylabel="AUC", title="Разброс качества по фолдам", ylim=(0, 1))

plt.tight_layout()
plt.show()
```

```python
frequency_comparison = pd.DataFrame({
    "фактическая доля дефолтов": [y.mean()],
    "средний OOF PD": [pd_oof.mean()],
}, index=["Logit"])

display(frequency_comparison.round(5))
```

```python
fold_metrics = []
for _, row in folds.iterrows():
    mask = oof["outer_fold"].to_numpy() == row["fold"]
    fold_y, fold_pd = y[mask], pd_oof[mask]
    fold_fpr, fold_tpr, _ = roc_curve(fold_y, fold_pd)
    fold_auc = roc_auc_score(fold_y, fold_pd)
    baseline_pd = row["train_defaults"] / row["train_n"]
    fold_r2 = 1 - log_loss(fold_y, fold_pd, normalize=False) / log_loss(
        fold_y, np.full(len(fold_y), baseline_pd), normalize=False
    )
    fold_metrics.append({
        "fold": int(row["fold"]),
        "AUC": fold_auc,
        "Gini": 2 * fold_auc - 1,
        "KS": np.max(fold_tpr - fold_fpr),
        "OOF псевдо-R²": fold_r2,
    })

fold_metrics = pd.DataFrame(fold_metrics).set_index("fold")
null_levels = {"AUC": 0.5, "Gini": 0.0, "KS": 0.0, "OOF псевдо-R²": 0.0}
stat_rows = []

for metric, null in null_levels.items():
    values = fold_metrics[metric].to_numpy()
    se = values.std(ddof=1) / np.sqrt(len(values))
    t_stat = (values.mean() - null) / se
    stat_rows.append({
        "метрика": metric,
        "среднее по фолдам": values.mean(),
        "стандартная ошибка": se,
        "t": t_stat,
        "p-value (прибл.)": student_t.sf(t_stat, df=len(values) - 1),
    })

stat_significance = pd.DataFrame(stat_rows).set_index("метрика")
display(stat_significance.round(5))
print("Проверочный псевдо-R² по фолдам:", fold_metrics["OOF псевдо-R²"].round(4).tolist())
```

```python
final_fit = fit_score_model(df)
final_eta = linear_score(final_fit, df)
final_pd = expit(final_eta)

print("Отбор на полном наборе:", final_fit["features"])
print("Причины исключения:", final_fit["history"])
print(f"Псевдо-R² Макфаддена на train: {final_fit['model'].prsquared:.4f} (описательно)")

display(final_fit["screen"])

coef = pd.DataFrame({
    "коэффициент": final_fit["model"].params,
    "cluster-robust SE": final_fit["model"].bse,
    "cluster-robust p-value": final_fit["model"].pvalues,
})
display(coef.round(5))

terms = " ".join(
    f"{v:+.5f}·({name})"
    for name, v in final_fit["model"].params.items()
    if name != "const"
)
print(f"eta = {final_fit['model'].params['const']:.5f} {terms}")
print("PD = 1 / (1 + exp(-eta))")
print(f"Диапазон итоговых PD: {final_pd.min():.3g}–{final_pd.max():.5f}; средний {final_pd.mean():.3%}")
```
